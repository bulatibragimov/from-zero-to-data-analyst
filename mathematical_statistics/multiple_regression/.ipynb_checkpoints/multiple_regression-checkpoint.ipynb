## Создание признака `company`

Использовать полное название автомобиля не всегда удобно, поэтому создадим новый признак — марку автомобиля (`company`).

Для этого используйте столбец `CarName`, разделите значения по пробелу и сохраните первый элемент в новую колонку.

Пример:

```text
'audi 100 ls' → 'audi'
```

После создания нового признака:

1. Определите количество уникальных марок автомобилей в датасете.
2. Проверьте полученные значения и при необходимости исправьте ошибки в названиях производителей.
3. Удалите столбцы:
   - `CarName`
   - `car_ID`

Так как они не потребуются для дальнейшего анализа.

### Вопрос

````python
cars['company'] = cars.CarName.str.split(' ').str[0]
````

## Очистка названий производителей

После создания признака `company` оказалось, что некоторые названия производителей содержат ошибки и дубликаты.

Исправьте значения следующим образом:

```text
'maxda'      → 'mazda'
'Nissan'     → 'nissan'
'porcshce'   → 'porsche'
'toyouta'    → 'toyota'
'vokswagen'  → 'volkswagen'
'vw'         → 'volkswagen'
```

После внесения исправлений повторно проверьте количество уникальных производителей.

### Вопрос

Сколько уникальных производителей автомобилей осталось в датасете после исправления ошибок?

````python
cars.company.unique()
````

````python
cars['company'] = cars.company.replace(to_replace={'maxda':'mazda', 'Nissan':'nissan', 'porcshce':'porsche', 'toyouta':'toyota','vokswagen':'volkswagen','vw' :'volkswagen'})
````

````python
cars.company.nunique()
````

````python
cols_to_keep = ['price', 'company', 'fueltype', 'aspiration','carbody', 
'drivewheel', 'wheelbase', 'carlength','carwidth', 
'curbweight', 'enginetype', 'cylindernumber', 'enginesize', 
'boreratio','horsepower']
````

````python
cars.corr(numeric_only=True)
````

````python
df_dummy = pd.get_dummies(data=cars[cols_to_keep], drop_first=True)
````

````python
df_dummy.head()
````

````python
import statsmodels.formula.api as smf
formula = 'price ~ horsepower'
smf.ols(formula, df_dummy).fit().summary()
````

````python
import statsmodels.api as sm
````

````python
cols_to_keep_without_price = ['company', 'fueltype', 'aspiration','carbody', 
'drivewheel', 'wheelbase', 'carlength','carwidth', 
'curbweight', 'enginetype', 'cylindernumber', 'enginesize', 
'boreratio','horsepower']
````

````python
X = df_dummy.drop(columns='price')
y = df_dummy['price']
````

````python
df_dummy = df_dummy.astype(float)
````

````python
model_sklrn = sm.OLS(endog=y, exog=X).fit()
model_sklrn.summary()
````

````python
X_1 = df_dummy[['wheelbase','carlength','carwidth','curbweight','enginesize','boreratio','horsepower']]
````

````python
model_sklrn = sm.OLS(endog=y, exog=X_1).fit()
model_sklrn.summary()
````

````python
df_without_company = df_dummy.loc[:, ~df_dummy.columns.str.startswith('company_')]

X = df_without_company.drop(columns='price')
y = df_without_company['price']

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
model.summary()
````
