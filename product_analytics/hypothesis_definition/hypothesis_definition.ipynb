# Приоритизация гипотез EasyRead по ICE

## Исходные данные

Всего пользователей: **12 000 000**

Платящих пользователей: **3 000 000**

Неплатящих пользователей:

```python
non_paying_users = 12_000_000 - 3_000_000
non_paying_users = 9_000_000
```

Текущая конверсия в подписчиков:

```python
conversion = 3_000_000 / 12_000_000 = 25%
```

Для расчёта ICE используется формула:

```text
ICE = Impact × Confidence × Ease
```

Где:

* **Impact** — прирост платящих пользователей;
* **Confidence** — уверенность в гипотезе по шкале от 1 до 10;
* **Ease** — простота реализации по шкале от 1 до 10.

---

impact_1 = 9_000_000 * 0.2 * 0.3

**Impact гипотезы 2: 72 000 новых платящих пользователей**

## Масштабирование Impact в шкалу от 1 до 10



Нужно перевести значения в шкалу от **1 до 10** по формуле линейного масштабирования:

$$
x_{scaled} = a + (x - x_{min}) \cdot \frac{b - a}{x_{max} - x_{min}}
$$

где:

- $x$ — исходное значение Impact;
- $x_{min}$ — минимальное значение Impact;
- $x_{max}$ — максимальное значение Impact;
- $a = 1$ — нижняя граница новой шкалы;
- $b = 10$ — верхняя граница новой шкалы.



````python
impact_values = {
    "Гипотеза 1": 540_000,
    "Гипотеза 2": 72_000,
    "Гипотеза 3": 150_000,
    "Гипотеза 4": 300_000,
    "Гипотеза 5": 450_000
}

x_min = min(impact_values.values())
x_max = max(impact_values.values())

a = 1
b = 10

scaled_impact = {}

for hypothesis, x in impact_values.items():
    x_scaled = a + (x - x_min) * (b - a) / (x_max - x_min)
    scaled_impact[hypothesis] = int(x_scaled)

scaled_impact
````

````python
import pandas as pd

impact_values = {
    "Рекомендации": non_paying_users * 0.2 * 0.3,
    "Ускорение загрузки": total_users * 0.3 * 0.02,
    "Тексты в email-рассылках": paying_users * 0.05,
    "Пробный период с расширенным функционалом": paying_users * 0.10,
    "Тариф по сниженной цене": non_paying_users * 0.05
}

confidence_values = {
    "Рекомендации": 4,
    "Ускорение загрузки": 3,
    "Тексты в email-рассылках": 7,
    "Пробный период с расширенным функционалом": 9,
    "Тариф по сниженной цене": 2
}

ease_values = {
    "Рекомендации": 5,
    "Ускорение загрузки": 4,
    "Тексты в email-рассылках": 9,
    "Пробный период с расширенным функционалом": 7,
    "Тариф по сниженной цене": 2
}

df = pd.DataFrame({
    "hypothesis": impact_values.keys(),
    "impact_raw": impact_values.values()
})

x_min = df["impact_raw"].min()
x_max = df["impact_raw"].max()

a = 1
b = 10

df["impact"] = (
    a + (df["impact_raw"] - x_min) * (b - a) / (x_max - x_min)
).astype(int)

df["confidence"] = df["hypothesis"].map(confidence_values)
df["ease"] = df["hypothesis"].map(ease_values)

df["ICE"] = df["impact"] * df["confidence"] * df["ease"]

df.sort_values("ICE", ascending=False)
````
