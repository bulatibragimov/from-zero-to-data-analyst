# A/B-тест нового алгоритма в приложении для онлайн-знакомств

## Описание задачи

Вы работаете аналитиком в приложении для онлайн-знакомств.

Механика приложения устроена следующим образом:

* пользователи видят анкеты друг друга;
* пользователь может поставить анкете **лайк** или **дизлайк**;
* если два пользователя взаимно поставили друг другу лайк, происходит **мэтч**;
* после мэтча у пользователей появляется возможность познакомиться.

## Эксперимент

Команда приложения разработала новый алгоритм подбора наиболее подходящих анкет.

Чтобы проверить эффективность нового алгоритма, был проведён A/B-тест. Все пользователи были разделены на две группы:

* **Группа 0** — пользователи со старым алгоритмом подбора анкет;
* **Группа 1** — пользователи с новым алгоритмом подбора анкет.

## Данные

В данных содержится выгрузка логов взаимодействий пользователей друг с другом.

Для каждой пары пользователей указаны:

* группа в A/B-тесте;
* факт мэтча между пользователями.

## Задача аналитика

Необходимо оценить, действительно ли новый алгоритм улучшил качество сервиса.

Для этого нужно:

1. выбрать метрики, которые отражают качество работы сервиса;
2. рассчитать эти метрики для двух групп;
3. статистически сравнить значения метрик между группами;
4. сделать вывод о влиянии нового алгоритма.

## Итоговый результат

Результатом работы должно быть аналитическое заключение с ответом на вопрос:

**Стоит ли включать новую систему поиска анкет на всех пользователей?**


## Кейс 1
Чему равна средняя доля мэтчей на пользователя в 0 группе? Ответ нужно предоставить до 4-х знаков после запятой без округлений.

```python
import pandas as pd
```

```python
dating_data = pd.read_csv('dating_data.csv')
```

```python
dating_data.head()
```

```python
average_share_0 = dating_data.query('group == 0').groupby('user_id_1').is_match.mean().round(2).mean()
average_share_0
```

## Кейс 2
Чему равна средняя доля мэтчей на пользователя в 1 группе? Ответ нужно предоставить до 4-х знаков после запятой без округлений.



```python
average_share_1 = dating_data.query('group == 1').groupby('user_id_1').is_match.mean().round(2).mean()
average_share_1
```

## Кейс 3
Проверьте, есть ли различия в средней доле мэтчей на пользователя в двух полученных группах. 



```python
test = dating_data.query('group == 1').groupby('user_id_1', as_index=False).agg(match_share=('is_match', 'mean')).match_share.round(2)
control = dating_data.query('group == 0').groupby('user_id_1', as_index=False).agg(match_share=('is_match', 'mean')).match_share.round(2)
```

```python
import scipy.stats as st

st.ttest_ind(control, test, equal_var=False)
```

Для проверки гипотезы я использую T-КРИТЕРИЙ УЭЛЧА,
поскольку переменные обе — КОЛИЧЕСТВЕННЫЕ.
Нулевая гипотеза ОТКЛОНЯЕТСЯ,
поскольку p-value МЕНЬШЕ 0.05.

## Кейс 4

## Кейс 9

```python
testing = testing = dating_data.query('group == 1').groupby('user_id_1').agg( count_1 = ( 'is_match' ,'count')).count_1
controling = dating_data.query('group == 0').groupby('user_id_1').agg( count_0 = ( 'is_match' ,'count')).count_0
```
